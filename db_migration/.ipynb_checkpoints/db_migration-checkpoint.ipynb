```python
from dataclasses import dataclass
import pandas as pd
from sqlalchemy import (
    create_engine,
    text,
    MetaData,
    Table,
)
from sqlalchemy.dialects.postgresql import insert
import os
```

```python
# CONFIGURATION
DB_URL = ""  # fill e.g postgresql://<user>:root@localhost:<port>/<dbname> default port: 5432, example without auth
engine = create_engine(DB_URL)
metadata = MetaData()
```

```python
# CREATE TABLES FROM SCHEMAS
# Keep quotation around names (else placeId -> placeid after migration)

table_schema_places = """
CREATE TABLE IF NOT EXISTS places (
    "placeId" INTEGER PRIMARY KEY,
    "lat" DOUBLE PRECISION NOT NULL,
    "lon" DOUBLE PRECISION NOT NULL,
    "placeName" VARCHAR(255) NOT NULL,
    "placeType" VARCHAR(255),
    "address" VARCHAR(255),
    "website" VARCHAR(255),
    "phone" VARCHAR(50),
    "note" VARCHAR(255),
    "validUntil" DATE
);
"""

table_schema_products = """
CREATE TABLE IF NOT EXISTS products (
    "productId" INTEGER PRIMARY KEY,
    "brandName" VARCHAR(255) NOT NULL,
    "productName" VARCHAR(255) NOT NULL,
    "productType" VARCHAR(255) NOT NULL
);
"""

table_schema_entries = """
CREATE TABLE IF NOT EXISTS entries (
    "entryId" INTEGER PRIMARY KEY,
    "placeId" INTEGER NOT NULL,
    "productId" INTEGER NOT NULL,
    "price" FLOAT,
    "volume" FLOAT,
    "vomFass" BOOLEAN,
    "validFrom" DATE,
    "lastUpdate" DATE,
    "validUntil" DATE,
    FOREIGN KEY ("placeId") REFERENCES places ("placeId")
        ON UPDATE CASCADE ON DELETE CASCADE,
    FOREIGN KEY ("productId") REFERENCES products ("productId")
        ON UPDATE CASCADE ON DELETE CASCADE,
    CONSTRAINT unique_place_product_validfrom UNIQUE ("placeId", "productId", "validFrom")
);
"""

# Create Tables

with engine.connect() as conn:
    conn.execute(text(table_schema_places))
    conn.execute(text(table_schema_products))
    conn.execute(text(table_schema_entries))
```

```python
# LOAD CSVs

# PATHS
main_path = "/Users/steffi/PycharmProjects/bmap-web/public/"  # fill
path_places = main_path + "places.csv"
path_products = main_path + "products.csv"
path_entries = main_path + "entries.csv"

df_places = pd.read_csv(path_places)
df_places = df_places.where(pd.notnull(df_places), None)

df_products = pd.read_csv(path_products)

df_entries = pd.read_csv(path_entries)

# Replace empty strings and NaN/NaT → None (for SQL compatibility)
df_entries = df_entries.replace(" ", None)
df_entries = df_entries.where(pd.notnull(df_entries), None)


# UPSERT PLACES
places_table = Table("places", metadata, autoload_with=engine)

with engine.begin() as conn:
    for _, row in df_places.iterrows():
        stmt = insert(places_table).values(row.to_dict())
        stmt = stmt.on_conflict_do_update(
            index_elements=["placeId"],
            set_={
                col: getattr(stmt.excluded, col)
                for col in row.index
                if col != "placeId"
            },
        )
        conn.execute(stmt)

# UPSERT PRODUCTS
products_table = Table("products", metadata, autoload_with=engine)

with engine.begin() as conn:
    for _, row in df_products.iterrows():
        stmt = insert(products_table).values(row.to_dict())
        stmt = stmt.on_conflict_do_update(
            index_elements=["productId"],
            set_={
                col: getattr(stmt.excluded, col)
                for col in row.index
                if col != "productId"
            },
        )
        conn.execute(stmt)

# UPSERT ENTRIES
entries_table = Table("entries", metadata, autoload_with=engine)

with engine.begin() as conn:
    for _, row in df_entries.iterrows():
        stmt = insert(entries_table).values(row.to_dict())
        stmt = stmt.on_conflict_do_update(
            index_elements=["placeId", "productId", "validFrom"],
            set_={
                col: getattr(stmt.excluded, col)
                for col in row.index
                if col not in ("placeId", "productId", "validFrom")
            },
        )

        conn.execute(stmt)
```
